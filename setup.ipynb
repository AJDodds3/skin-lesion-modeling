# Setup and initial data check

For now I am just checking that the data loads and seeing whether missing clinical fields vary across diagnoses. 

```python
from pathlib import Path
import pandas as pd

file = Path("data/metadata.csv")

if file.exists():
    print("Using the full metadata file")
else:
    print("Data is missing")
data = pd.read_csv(file)
print("Rows:", len(data))
print("Patients:", data["patient_id"].nunique())
print("\nCases by diagnosis:")
print(data["diagnostic"].value_counts().sort_index())
```

```python
ignore = ["patient_id", "lesion_id", "img_id", "diagnostic", "biopsed"]
fields = []
for column in data.columns:
    if column not in ignore:
        fields.append(column)

# Count both blank fields and entries marked unknown.
missing = data[fields].isna() | data[fields].eq("UNK")
missing_counts = missing.groupby(data["diagnostic"]).sum()

print("Missing or unknown values by diagnosis:")
display(missing_counts)

if file.name == "metadata.csv":
    missing_counts.to_csv("results/missingness_by_diagnosis.csv")
    print("Saved the full results to results/missingness_by_diagnosis.csv")
```
